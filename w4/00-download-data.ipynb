```python
from playwright.async_api import async_playwright

## now let us start a browser
url="https://ocrportal.hhs.gov/ocr/breach/breach_report_hip.jsf"

p = await async_playwright().start()
## initiate browser
browser = await p.chromium.launch(headless=False)
## create a new page
page = await browser.new_page()

await page.goto(url)

### let us click on Archive
await page.get_by_role("link", name="Archive", exact=True).click()

async with page.expect_download() as download_info:
    await page.get_by_role("link", name="CSV", exact=True).click()

download= await download_info.value
await download.save_as("downloaded_file.csv")

await browser.close()
await p.stop()
```
